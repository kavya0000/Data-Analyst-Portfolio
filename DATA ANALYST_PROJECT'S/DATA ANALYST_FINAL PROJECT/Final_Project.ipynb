**DATA CLEANING & PREPARATION**

1.IMPORTING DATA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/healthcare_dataset.csv")
```

2. CLEANING  THE DATA

```python
#check for missing values
df.isnull().sum()
```

```python
#Fill missing values
df.fillna(0)
```

```python
#drop missing values
df.dropna()
```

```python
#remove duplicates
df.drop_duplicates(inplace=True)
```

3. DATA TRANSFORMATION

```python
print("Original Dataset:")
print(df.head())
```

```python
# Data Type Conversion

# Convert 'Date of Admission' and 'Discharge Date' to datetime
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], errors='coerce')
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'], errors='coerce')

# Convert 'Room Number' to string (if it should be treated as ID)
df['Room Number'] = df['Room Number'].astype(str)
```

```python
# Data Cleaning / Normalization

# Clean up 'Name' column – proper capitalization
df['Name'] = df['Name'].str.title().str.strip()

# Standardize text columns (Gender, Medical Condition, Hospital, Insurance Provider)
text_columns = ['Gender', 'Medical Condition', 'Hospital', 'Insurance Provider', 'Admission Type', 'Medication', 'Test Results']
for col in text_columns:
    df[col] = df[col].str.title().str.strip()
```

```python
# Create New Columns for Analysis
df['Admission_Year'] = df['Date of Admission'].dt.year
df['Admission_Month'] = df['Date of Admission'].dt.month_name()

# Calculate Length of Stay (in days)
df['Length_of_Stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

# Create Age Group column
df['Age_Group'] = pd.cut(df['Age'],
                         bins=[0, 12, 19, 35, 50, 65, 120],
                         labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Middle Age', 'Senior'])
```

```python
# Aggregation Examples

# Average billing amount by Gender
avg_billing_gender = df.groupby('Gender')['Billing Amount'].mean().round(2)
print("\nAverage Billing Amount by Gender:")
print(avg_billing_gender)

# Total number of patients by Year of Admission
patients_per_year = df.groupby('Admission_Year')['Name'].nunique()
print("\nTotal Patients per Year:")
print(patients_per_year)

# Average Length of Stay by Medical Condition
avg_stay_condition = df.groupby('Medical Condition')['Length_of_Stay'].mean().round(1)
print("\nAverage Length of Stay (days) by Medical Condition:")
print(avg_stay_condition)
```

```python
print("\n Data transformation completed successfully!")
print("Transformed data saved as 'healthcare_data_transformed.csv'")
```

2. EXPLORATORY DATA ANALYSIS

```python

# EXPLORATORY DATA ANALYSIS (EDA) - HEALTHCARE DATASET


import matplotlib.pyplot as plt
import seaborn as sns

# Histogram - Age Distribution
plt.hist(df['Age'], bins=10, color='skyblue', edgecolor='black')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Histogram - Billing Amount Distribution
plt.hist(df['Billing Amount'], bins=10, color='orange', edgecolor='black')
plt.title('Billing Amount Distribution')
plt.xlabel('Billing Amount')
plt.ylabel('Frequency')
plt.show()

# Bar Chart - Gender Count
sns.countplot(x='Gender', data=df, palette='pastel')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()

# Bar Chart - Medical Condition Count
sns.countplot(x='Medical Condition', data=df, palette='Set2')
plt.title('Medical Condition Distribution')
plt.xlabel('Medical Condition')
plt.ylabel('Count')
plt.show()

# Scatter Plot - Age vs Billing Amount
sns.scatterplot(x='Age', y='Billing Amount', hue='Gender', data=df)
plt.title('Age vs Billing Amount by Gender')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.show()

# Box Plot - Billing Amount by Medical Condition
sns.boxplot(x='Medical Condition', y='Billing Amount', data=df, palette='coolwarm')
plt.title('Billing Amount by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Billing Amount')
plt.show()
```

 DATA VISUALIZATION

```python

# Ensure 'Date of Admission' is datetime
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], errors='coerce')

# 1️⃣ Line Plot - Average Billing Amount by Year of Admission
avg_billing_year = df.groupby(df['Date of Admission'].dt.year)['Billing Amount'].mean()
avg_billing_year.index.name = 'Year'

plt.figure()
avg_billing_year.plot(kind='line', marker='o', color='teal')
plt.title('Average Billing Amount by Year of Admission')
plt.xlabel('Year of Admission')
plt.ylabel('Average Billing Amount')
plt.grid(True)
plt.show()

# 2️⃣ Bar Chart - Average Billing Amount by Medical Condition
avg_billing_condition = df.groupby('Medical Condition')['Billing Amount'].mean().sort_values(ascending=False)
plt.figure()
avg_billing_condition.plot(kind='bar', color='coral')
plt.title('Average Billing Amount by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Billing Amount')
plt.xticks(rotation=45, ha='right')
plt.show()

# 3️⃣ Pie Chart - Gender Distribution
gender_counts = df['Gender'].value_counts()
plt.figure()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Gender Distribution')
plt.axis('equal')
plt.show()

# 4️⃣ Heatmap - Correlation Between Numeric Columns
numeric_cols = ['Age', 'Billing Amount']
if 'Room Number' in df.columns:
    numeric_cols.append('Room Number')  # include Room Number if numeric

plt.figure(figsize=(6,4))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Billing Amount according to Medical Condition and Medication
df_trans = df.groupby(['Medical Condition', 'Medication'])[['Billing Amount']].sum().reset_index()

plt.figure(figsize=(15,6))
sns.barplot(x=df_trans['Medical Condition'], y=df_trans['Billing Amount'], hue=df_trans['Medication'], ci=None, palette="Set1")
plt.title("Billing Amount according to Medical Condition and Medication")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45, fontsize = 9)
plt.show()
```
